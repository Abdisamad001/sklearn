```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('/content/housing.csv')
```

```python
df = df.drop('ocean_proximity', axis = 1)
df = df.dropna()
```

```python
X = df.drop('median_house_value', axis = 1)
y = df['median_house_value']
```

```python
Xtrain, Xtest, ytrain, ytest = train_test_split(X, y, test_size=0.2)
```

```python
print("Xtrain: ", Xtrain.shape)
print("ytrain: ", ytrain.shape)
print("Xtest: ", Xtest.shape)
print("ytest: ", ytest.shape)
```

# Linear Regression from scratch (Numpy)

```python
A_train = Xtrain.to_numpy()
A_test = Xtest.to_numpy()
train_target = ytrain.to_numpy()
test_target = ytest.to_numpy()
```

```python
theta = np.linalg.solve(A_train.T @ A_train, A_train.T @ train_target.reshape(-1,1))
yhat_train = A_train @ theta
yhat_test= A_test @ theta
```

```python
print(f'Train MSE: {mean_squared_error(yhat_train, train_target)}')
print(f'Test MSE: {mean_squared_error(yhat_test, test_target)}')
```

# Linear Regression using (Sklearn)

```python
model = LinearRegression()
```

```python
model.fit(Xtrain, ytrain)
```

```python
yhat_train = model.predict(Xtrain)
yhat_test = model.predict(Xtest)
```

```python
train_loss = mean_squared_error(yhat_train, ytrain)
test_loss = mean_squared_error(yhat_test, ytest)
```
